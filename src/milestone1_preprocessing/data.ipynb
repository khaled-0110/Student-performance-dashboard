```python
import pandas as pd
students=pd.read_csv('students.csv')
courses=pd.read_csv('courses.csv')
attendance=pd.read_csv('attendance.csv')
grades=pd.read_csv('grades.csv')
```

```python
df=attendance.merge(students,on="StudentID")
df=df.merge(courses, on="CourseID")
df=df.merge(grades,on="StudentID")
print(df.columns)
```

```python
print(df)
```

```python
final_df=df[["StudentID","FirstName","Department","Score","Date_y","AttendanceStatus"]]
print(final_df.columns)
```
