```python
#load data
import pandas as pd
data=pd.read_csv("raw_students.csv")
data
```

```python
#show duplicates numbers
data.duplicated().sum()
```

```python
#show duplicates in data
data[data.duplicated()]
```

```python
#remove duplicates
data.drop_duplicates(inplace=True)
```

```python
#check about duplicates
data.duplicated().sum()
```

```python
#convert dates from object to date
data['date']=pd.to_datetime(data['date'])
```

```python
data.info()
```

```python
#show number of nulls
data.isnull().sum()
```

```python
#fix nulls 
subject_means = data.groupby('subject')['score'].transform('mean')
data['score'] = data['score'].fillna(subject_means)

data['score'] = data['score'].fillna(data['score'].mean())

data['attendance'] = data['attendance'].fillna('Unknown')

data['subject'] = data['subject'].fillna('Unknown')

data['name'] = data['name'].fillna('Unknown')

data['date']=pd.to_datetime(data['date'],errors='coerce')
data['date']=data['date'].fillna(pd.to_datetime('today'))
```

```python
#check about nulls
data.isnull().sum()
```

```python
#categorize performance 
def categorize_performance(score):
    if score >=85:
        return "High"
    elif score >=60:
        return "Medium"
    else:
        return "Low"
```

```python
#add now column(categorize performance )
data['categorize_performance']=data['score'].apply(categorize_performance) 
```

```python
data.to_csv('cleaned_students.csv',index=False)
```
